```python
import pandas as pd
import numpy as np

data = [
["mouse","Logitech",120,"preto","entrada"],
["mouse","Razer",350,"preto","premium"],
["mouse","Multilaser",40,"preto","entrada"],
["teclado","Redragon",280,"preto","intermediario"],
["teclado","Multilaser",60,"branco","entrada"],
["teclado","Corsair",600,"preto","premium"],
["headset","Multilaser",80,"preto","entrada"],
["headset","HyperX",450,"preto","premium"],
["headset","Logitech",220,"preto","intermediario"],
["monitor","Samsung",700,"preto","intermediario"],
["monitor","Asus",1800,"preto","premium"],
["monitor","AOC",500,"preto","entrada"],
["mousepad","Fortrek",30,"preto","entrada"],
["mousepad","Razer",200,"preto","intermediario"],
["webcam","Logitech",250,"preto","intermediario"],
["webcam","Microsoft",400,"preto","premium"],
["cadeira","DXRacer",900,"preto","intermediario"],
["cadeira","Secretlab",2500,"preto","premium"],
["som","Multilaser",70,"preto","entrada"],
["som","JBL",300,"azul","intermediario"],
]

cols = ["categoria","marca","preco","cor","tipo"]

df = pd.DataFrame(data, columns=cols)
```

```python
TARGET = "tipo"
atributos = ["categoria","marca","cor"]

classes = df[TARGET].unique()
total = len(df)
```

```python
prob_classe = {}

for c in classes:
    prob_classe[c] = (len(df[df[TARGET] == c]) + 1) / (total + len(classes))

print("Probabilidade das classes:")
print(prob_classe)
```

```python
def prob_condicional(df, atributo, classe):
    valores = df[atributo].unique()
    k = len(valores)

    probs = {}

    df_classe = df[df[TARGET] == classe]
    total_classe = len(df_classe)

    for v in valores:
        count = len(df_classe[df_classe[atributo] == v])
        probs[v] = (count + 1) / (total_classe + k)

    return probs

tabelas = {}

for c in classes:
    tabelas[c] = {}
    for a in atributos:
        tabelas[c][a] = prob_condicional(df, a, c)
```

```python
produto_novo = {
    "categoria": "mouse",
    "marca": "Logitech",
    "cor": "preto"
}
```

```python
resultados = {}

for c in classes:
    prob = prob_classe[c]

    for a in atributos:
        prob *= tabelas[c][a][produto_novo[a]]

    resultados[c] = prob

soma = sum(resultados.values())
for c in resultados:
    resultados[c] /= soma

print("\nProbabilidades finais:")
print(resultados)

classe_prevista = max(resultados, key=resultados.get)

print("\nClasse prevista:", classe_prevista)
```
